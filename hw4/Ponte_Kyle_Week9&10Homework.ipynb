```python
class Point():
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __sub__(self, other):
        return Point(self.x - other.x, self.y - other.y) # Subtract two points

    def __str__(self):
        return f'({self.x}, {self.y})' # String representation of the point

p1 = Point(5, 7)
p2 = Point(2, 3)
p3 = p1 - p2
print(p3)
```

```python
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def __getattr__(self, name):
        if name == 'area':
            return 3.14159 * (self.radius ** 2) # Calculate area of the circle
        elif name == 'circumference':
            return 2 * 3.14159 * self.radius # Calculate circumference of the circle
        else:
            raise AttributeError(f"'Circle' object has no attribute '{name}'")

c = Circle(3)
print(c.area)
print(c.circumference)
```

```python
'''
In object oriented programming, polymorphism allows different classes to define methods with the same name, but potentially different behaviors. When the method is called, Python automatically chooses the correct version based on the object type. This makes code more flexible and easier to extend. Encapsulation is the concept of hiding internal data and requiring interaction with an object only through its methods. It protects the internal state of an object from direct modification. In Python, we indicate private data or methods by starting their names with an underscore.
'''

class BankAccount:
    def __init__(self, owner, balance):
        # Encapsulation: balance is protected and cannot be changed directly
        self._owner = owner
        self._balance = balance

    # Method to safely view balance
    def get_balance(self):
        return self._balance

    # Method that will be overridden (polymorphism)
    def account_type(self):
        return "Generic Bank Account"

    # Method to safely modify balance
    def deposit(self, amount):
        self._balance += amount


class SavingsAccount(BankAccount):
    def __init__(self, owner, balance, interest_rate):
        super().__init__(owner, balance)
        self._interest_rate = interest_rate

    # Polymorphism: overriding the parent method
    def account_type(self):
        return "Savings Account with interest"

    def add_interest(self):
        self._balance += self._balance * self._interest_rate


acct1 = BankAccount("Kyle", 500)
acct2 = SavingsAccount("John", 500, 0.10)

print(acct1.account_type())
print(acct1.get_balance())

print(acct2.account_type())
acct2.add_interest()
print(acct2.get_balance())

```

```python
class Employee:
    def __init__(self, name, hourly_rate):
        self.name = name
        self._hourly_rate = hourly_rate # Protected attribute

    # Encapsulation: Getter
    def get_hourly_rate(self):
        return self._hourly_rate

    # Encapsulation: Setter with validation
    def set_hourly_rate(self, new_rate):
        if new_rate < 0:
            print("Error: Hourly rate cannot be negative.")
        else:
            self._hourly_rate = new_rate

    # Method that will be overridden (polymorphism)
    def calculate_pay(self, hours):
        return hours * self._hourly_rate


class Manager(Employee):  # Manager inherits from Employee
    def __init__(self, name, hourly_rate, bonus):
        super().__init__(name, hourly_rate)
        self._bonus = bonus

    # Polymorphism: override calculate_pay
    def calculate_pay(self, hours):
        return (hours * self._hourly_rate) + self._bonus


e1 = Employee("Kyle", 20)
m1 = Manager("John", 25, 200)

print("Employee pay:", e1.calculate_pay(40))
print("Manager pay:", m1.calculate_pay(40))

e1.set_hourly_rate(22) # safe update
print("Updated employee pay:", e1.calculate_pay(40))

e1.set_hourly_rate(-5) # invalid update
```
